```python
import numpy as np

def evaluation(file_path):
    pre_list = []
    gt_list = []

    with open(file_path, 'r') as file:
        for line in file:
            line = line.strip()
            line_list = line.split(',')
            pre = line_list[0].replace('[','').replace(']','').split(' ')
            pre_2 = []
            for i in pre:
                if i != '':
                    pre_2.append(float(i))

            
            gt = line_list[1].replace('[','').replace(']','').split(' ')
            gt = [float(i) for i in gt]
            gt_list.append(gt)
            max_indices = np.argmax(pre_2) 
            pre = [0, 0, 0, 0, 0] # 形状与prob相同
            pre[max_indices] = 1
            pre_list.append(pre)

    gt_list_count = []
    pre_list_count = []
    for i in range(len(gt_list)):
        gt_indices = np.argmax(gt_list[i]) 
        pre_indices = np.argmax(pre_list[i]) 
        gt_list_count.append(gt_indices)
        pre_list_count.append(pre_indices)
        # print(gt_list[i], pre_list[i], gt_indices, pre_indices)

    from sklearn.metrics import confusion_matrix
    cm = confusion_matrix(gt_list_count, pre_list_count)

    # 初始化指标列表
    accuracies, sensitivities, specificities, precisions, f1_scores = [], [], [], [], []

    # 总样本数
    total_samples = np.sum(cm)

    # 遍历每个类别
    for c in range(len(cm)):
        TP = cm[c, c]  # True Positives
        FN = np.sum(cm[c, :]) - TP  # False Negatives
        FP = np.sum(cm[:, c]) - TP  # False Positives
        TN = total_samples - (TP + FP + FN)  # True Negatives

        # 计算各指标
        accuracy = (TP + TN) / total_samples
        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0
        precision = TP / (TP + FP) if (TP + FP) > 0 else 0
        f1_score = 2 * precision * sensitivity / (precision + sensitivity) if (precision + sensitivity) > 0 else 0

        # 添加到对应列表
        accuracies.append(accuracy)
        sensitivities.append(sensitivity)
        specificities.append(specificity)
        precisions.append(precision)
        f1_scores.append(f1_score)

    # 计算宏平均
    macro_accuracy = np.mean(accuracies)
    macro_sensitivity = np.mean(sensitivities)
    macro_specificity = np.mean(specificities)
    macro_precision = np.mean(precisions)
    macro_f1_score = np.mean(f1_scores)

    print('{:.4f} & {:.4f} & {:.4f} & {:.4f} & {:.4f}'.format(macro_accuracy, macro_sensitivity, macro_specificity, macro_precision, macro_f1_score))


file_path = '/workspace/Zero_JSW_Generation/experiments/Exp_downstream/SvdH_evaluation/results/SvdH_summary_wo_pre_100_vit.txt'
evaluation(file_path)

file_path = '/workspace/Zero_JSW_Generation/experiments/Exp_downstream/SvdH_evaluation/results/SvdH_summary_w_pre_100_vit.txt'
evaluation(file_path)
```
