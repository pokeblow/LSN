```python
file_path = '/workspace/Zero_JSW_Generation/experiments/Exp_downstream/JSW_evaluation/results/JSW_summary_w_pre_100_vit.txt'
pre_list = []
gt_list = []

with open(file_path, 'r') as file:
    for line in file:
        line = line.strip()
        line_list = line.split(',')
        pre_list.append(float(line_list[0]))
        gt_list.append(float(line_list[1]))

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, explained_variance_score

print('MSE & MAE & EVS & R2')
print('{:.4f} & {:.4f} & {:.4f} & {:.4f}'.format(mean_squared_error(pre_list, gt_list), mean_absolute_error(pre_list, gt_list), explained_variance_score(gt_list, pre_list), r2_score(gt_list, pre_list)))
```
